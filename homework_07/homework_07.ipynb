### Создайте одномерный массив размера 10, заполненный нулями и пятым элемент равным 1. Трансформируйте в двумерный массив.

### Создайте одномерный массив со значениями от 10 до 49 и разверните его (первый элемент становится последним). Найдите в нем все четные элементы.

```python
arr = np.arange(10, 50)
print(arr)
```

```python
reversed_arr = np.flip(arr)
print(reversed_arr)
```

```python
even_numbers = arr[arr % 2 == 0]
print(even_numbers)
```

### Создайте двумерный массив 3x3 со значениями от 0 до 8

```python
array = np.arange(9).reshape(3, 3)
print(array)
```

### Создайте массив 4x3x2 со случайными значениями. Найти его минимум и максимум.

```python
random_array = np.random.randint(0, 100, size=(4,3,2))
print(random_array)
```

```python
print(np.min(random_array))
print(np.max(random_array))
```

### Создайте два двумерных массива размерами 6x4 и 4x3 и произведите их матричное умножение. 

```python
A = np.random.rand(6, 4)
B = np.random.rand(4, 3)
C = np.dot(A, B)
print(A)
print(B)
print(C)
```

### Создайте случайный двумерный массив 7x7, найти у него среднее и стандартное оклонение. Нормализуйте этот массив.

```python
array = np.random.rand(7, 7)
print(array)
```

```python
mean_array = np.mean(array)
std_array = np.std(array)
print(mean_array)
print(std_array)
```

```python
arr = array * 10
normalized = (arr - np.min(arr)) / (np.max(arr) - np.min(arr))
print(normalized)
```

# III. Visualization

### Постройте гистограмму распределения признака 'total_bill'

```python
sns.histplot(data=tips, x='total_bill', bins=30)

plt.title('Распределение total_bill')
plt.xlabel('Сумма счёта ($)')
plt.ylabel('Количество наблюдений')
plt.show()
```

### Постройте scatterplot, представляющий взаимосвязь между признаками 'total_bill' и 'tip'

```python
sns.scatterplot(data=tips, x='total_bill', y='tip')   

plt.title('Зависимость чаевых от суммы счёта')
plt.xlabel('Сумма счёта ($)')
plt.ylabel('Чаевые ($)')

plt.show()
```

### Постройте pairplot

```python
sns.pairplot(tips)
plt.show()
```

### Постройте график взаимосвязи между признаками 'total_bill' и 'day'

```python
plt.figure(figsize=(9, 6))

sns.boxplot(data=tips, x='day', y='total_bill', 
            order=['Thur', 'Fri', 'Sat', 'Sun'])  

plt.title('Распределение суммы счёта по дням недели')
plt.xlabel('День недели')
plt.ylabel('Сумма счёта ($)')
plt.show()
```

### Постройте две гистограммы распределения признака 'tip' в зависимости от категорий 'time'

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.histplot(data=tips[tips['time'] == 'Lunch'], 
             x='tip', 
             bins=15, 
             color='skyblue', 
             ax=axes[0])
axes[0].set_title('Чаевые в обед (Lunch)')
axes[0].set_xlabel('Сумма чаевых ($)')

sns.histplot(data=tips[tips['time'] == 'Dinner'], 
             x='tip', 
             bins=15, 
             color='salmon', 
             ax=axes[1])
axes[1].set_title('Чаевые на ужине (Dinner)')
axes[1].set_xlabel('Сумма чаевых ($)')

plt.tight_layout()
plt.show()
```

### Постройте два графика scatterplot, представляющих взаимосвязь между признаками 'total_bill' и 'tip' один для Male, другой для Female и раскрасьте точки в зависимоти от признака 'smoker'

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.scatterplot(data=tips[tips['sex'] == 'Male'], 
                x='total_bill',
                y='tip',
                hue='smoker', palette={'Yes': 'red', 'No': 'green'},
                style='smoker', markers={'Yes': 'X', 'No': 'o'},
                s=30, 
                ax=axes[0])
axes[0].set_title('Чаевые от мужчин (Male)')
axes[0].set_xlabel('Сумма счета ($)') 
axes[0].set_ylabel('Чаевые ($)')
axes[0].legend(title='Курящий') 

sns.scatterplot(data=tips[tips['sex'] == 'Female'], 
                x='total_bill',
                y='tip',
                hue='smoker', palette={'Yes': 'red', 'No': 'green'},
                style='smoker', markers={'Yes': 'X', 'No': 'o'},
                s=30, 
                ax=axes[1])
axes[1].set_title('Чаевые от женщин (Female)')
axes[1].set_xlabel('Сумма счета ($)') 
axes[1].set_ylabel('Чаевые ($)')
axes[1].legend(title='Курящая') 

plt.suptitle('Зависимость чаевых от суммы счета (по полу и статусу курения)')
plt.tight_layout()
plt.show()
```
