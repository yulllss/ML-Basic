# I. Numpy

### Импортируйте NumPy

```python
import numpy as np
```

### Создайте одномерный массив размера 10, заполненный нулями и пятым элемент равным 1. Трансформируйте в двумерный массив.

```python
array = np.zeros(10)
array[4] = 1
print(array)
```

```python
array_2 = array.reshape(2, 5)
print(array_2)
```

### Создайте одномерный массив со значениями от 10 до 49 и разверните его (первый элемент становится последним). Найдите в нем все четные элементы.

```python
arr = np.arange(10, 50)
print(arr)
```

```python
reversed_arr = np.flip(arr)
print(reversed_arr)
```

```python
even_numbers = arr[arr % 2 == 0]
print(even_numbers)
```

### Создайте двумерный массив 3x3 со значениями от 0 до 8

```python
array = np.arange(9).reshape(3, 3)
print(array)
```

### Создайте массив 4x3x2 со случайными значениями. Найти его минимум и максимум.

```python
random_array = np.random.randint(0, 100, size=(4,3,2))
print(random_array)
```

```python
print(np.min(random_array))
print(np.max(random_array))
```

### Создайте два двумерных массива размерами 6x4 и 4x3 и произведите их матричное умножение. 

```python
A = np.random.rand(6, 4)
B = np.random.rand(4, 3)
C = np.dot(A, B)
print(A)
print(B)
print(C)
```

### Создайте случайный двумерный массив 7x7, найти у него среднее и стандартное оклонение. Нормализуйте этот массив.

```python
array = np.random.rand(7, 7)
print(array)
```

```python
mean_array = np.mean(array)
std_array = np.std(array)
print(mean_array)
print(std_array)
```

```python
arr = array * 10
normalized = (arr - np.min(arr)) / (np.max(arr) - np.min(arr))
print(normalized)
```

### Отберите строчки с 'total_bill' больше медианы и узнайте какой средний 'tip' в зависимости от 'sex'

```python
median_bill = tips['total_bill'].median()
print(median_bill)
high_bill = tips[tips['total_bill'] > median_bill]
tip_by_sex = high_bill.groupby('sex')['tip'].mean()
print(tip_by_sex)
```
